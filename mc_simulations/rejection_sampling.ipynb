```python
import numpy as np
from scipy.stats import gamma
import matplotlib.pyplot as plt
```

# Rejection Sampling

A method of generating samples of a target distribution by sampling from an easily generated proposal distribution

```python
exp_pdf = lambda x: np.exp(-x) # exp(1)

x_sample = np.linspace(0, 8, 100)
plt.plot(x_sample, exp_pdf(x_sample))
```

```python
num_samples = 1_000_000
X = np.random.uniform(0,8, size=(num_samples,))
Y = np.random.uniform(size=(num_samples,))
keep_mask = Y < exp_pdf(X)
```

```python
plt.scatter(X[keep_mask], Y[keep_mask], c="orange", label="Acceptance Region")
plt.scatter(X[~keep_mask], Y[~keep_mask], c="blue", label="Rejection Region")
plt.plot(x_sample, exp_pdf(x_sample), c="red", label="True PDF")
plt.legend()
plt.show()
```
